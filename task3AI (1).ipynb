 Constraint Satisfaction Problem (CSP)


```python
subjects = ["Maths", "Python", "DBMS", "AI"]

time_slots = ["9:00 AM", "11:00 AM", "2:00 PM"]

constraints = [
    ("Maths", "Python"),
    ("DBMS", "AI")
]

def is_valid(subject, time_slot, assignment):
    for s1, s2 in constraints:
        if subject == s1 and s2 in assignment:
            if assignment[s2] == time_slot:
                return False

        if subject == s2 and s1 in assignment:
            if assignment[s1] == time_slot:
                return False

    return True


def backtracking(assignment):
    if len(assignment) == len(subjects):
        return assignment

    for subject in subjects:
        if subject not in assignment:
            for time_slot in time_slots:
                if is_valid(subject, time_slot, assignment):
                    assignment[subject] = time_slot

                    result = backtracking(assignment)

                    if result is not None:
                        return result

                    del assignment[subject]

            return None


solution = backtracking({})

print("FINAL TIMETABLE")
print("----------------")

for subject in subjects:
    print(subject, "->", solution[subject])
```

Short Explanation
Variables: Maths, Python, DBMS, AI
Domains: 9:00 AM, 11:00 AM, 2:00 PM
Constraints: Maths ≠ Python, DBMS ≠ AI
Search: Backtracking
Each subject gets exactly one time slot.
If a constraint is violated, the program goes back and tries another slot.
